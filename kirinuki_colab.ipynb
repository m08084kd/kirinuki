# ✂️ kirinuki — YouTube の盛り上がりを自動で切り抜き

iPad やスマホのブラウザからでも使えます。**上から順に、各ブロックの左にある ▶ ボタンを押すだけ**です。

1. **① 準備**: ▶ を押す（最初の1回だけ・1分ほど）
2. **② 切り抜く**: URL を貼って ▶ を押す（動画の長さにより数分〜十数分）
3. **③ 保存する**: 保存先を選んで ▶ を押す

> 初めて ▶ を押すと「警告: このノートブックは Google が作成したものではありません」と出ます。「**このまま実行**」を押してください。

```python
#@title ① 準備（最初に1回だけ ▶ を押す）
KIRINUKI_REF = "claude/youtube-highlight-detection-clips-cemep2"
!pip install -q "git+https://github.com/m08084kd/kirinuki.git@{KIRINUKI_REF}"
from kirinuki import notebook
print("✅ 準備ができました。② に URL を入れて ▶ を押してください。")
```

```python
#@title ② 切り抜く（URL を貼って ▶ を押す）
URL = "" #@param {type:"string"}
本数 = 3 #@param {type:"slider", min:1, max:10, step:1}
最短の秒数 = 30 #@param {type:"integer"}
最長の秒数 = 60 #@param {type:"integer"}
前振りの秒数 = 10 #@param {type:"integer"}
縦型ショート = False #@param {type:"boolean"}
画質 = "720" #@param ["480", "720", "1080"]

from kirinuki import notebook
notebook.clip(URL, count=本数, min_len=最短の秒数, max_len=最長の秒数, lead_in=前振りの秒数,
              vertical=縦型ショート, max_height=int(画質))
```

```python
#@title ③ 保存する（保存先を選んで ▶ を押す）
保存先 = "Googleドライブ" #@param ["Googleドライブ", "ZIPでダウンロード"]

from kirinuki import notebook
notebook.save(保存先)
```

## うまくいかないとき

- **「Sign in to confirm you're not a bot」などのエラー**: YouTube に一時的に止められています。上のメニュー「ランタイム」→「ランタイムを接続解除して削除」を押してから、①からやり直してください。
- **しばらく放置したら動かない**: Colab は時間が経つと自動で止まります。①からやり直してください。
- **切り抜いた場面がずれている**: ② の「前振りの秒数」を増やすと、山場の前を長めに入れられます。
- **Google ドライブに保存した動画を写真アプリに入れたい**: iPad の「Google ドライブ」アプリで動画を開き、⋯ →「コピーを送信」→「ビデオを保存」。

切り抜きを公開する前に、元の配信者の切り抜きガイドラインを確認してください。
